```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import *
from sklearn.metrics import *
from sklearn.linear_model import *
import matplotlib.pyplot as plt
```

Reading data

```python
df = pd.read_csv("California_Houses.csv")
```

```python
df.nunique()
df.isnull().sum()
```

Dividing data into features and labels

```python
y=df.iloc[:,0]
X=df.iloc[:,1:14]
```

Splitting the data

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 100, shuffle = True)
X_val, X_test, y_val, y_test = train_test_split(X_test, y_test, test_size = 0.5, random_state = 100,shuffle=True)
```

Normalizing data

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
X_val=scaler.transform(X_val)
```

<h1>Linear regression</h1>

```python
linear_model=LinearRegression()
```

```python
linear_model.fit(X_train,y_train)
```

```python
y_preds_val=linear_model.predict(X_val)
y_preds=linear_model.predict(X_test)

# mse and mae for validation set
mse_val=mean_squared_error(y_preds_val,y_val)
mae_val=mean_squared_error(y_preds_val,y_val)



# mse and mae for test set
mse=mean_squared_error(y_preds,y_test)
mae=mean_squared_error(y_preds,y_test)
```

```python
print("Validation errors : ")
print(f"Mean Squared Error {mse_val} ")
print(f"Mean Absolute Error {mae_val} ")
print("---------------------------------------")
print("Test Errors : ")
print(f"Mean Squared Error {mse} ")
print(f"Mean Absolute Error {mae} ")
```

<h1>Ridge Regression</h1>
